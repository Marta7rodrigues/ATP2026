```python
#Aplicação das turmas de alunos
def guardar_turma_em_ficheiro(turma, nome_ficheiro):
    file = open(nome_ficheiro, "w")
    for aluno in turma:
        nome, id, notas = aluno
        file.write(f"{nome}#{id}#{notas[0]}#{notas[1]}#{notas[2]}\n")
    file.close()

def carregar_turma_de_ficheiro(nome_ficheiro):
    turma_carregada = []
    file = open(nome_ficheiro, "r")
    for linha in file:
        partes = linha.strip().split("#")
        nome = partes[0]
        id = int(partes[1])
        notas = [float(partes[2]), float(partes[3]), float(partes[4])]
        aluno = (nome, id, notas)
        turma_carregada.append(aluno)
    file.close()
    return turma_carregada

opção = ""
while opção != "0":
    print("Indique a operação que deseja realizar:")
    print("1: Criar uma turma; \n2: Inserir um aluno na turma; \n3: Listar a turma; \n4: Consultar um aluno por id; \n5: Guardar a turma em ficheiro; \n6: Carregar uma turma de um ficheiro; \n0: Sair")

    opção = input("Indique a operação que deseja realizar: ")

    if opção == "1":
        número_alunos = int(input("Indique o número de alunos da turma: "))
        turma = []
        while len(turma) < número_alunos:
            nome = input("Indique o nome do aluno: ")
            id = int(input("Indique o id do aluno: "))
            notaTPC = float(input("Indique a nota do TPC do aluno: "))
            notaProj = float(input("Indique a nota do Projeto do aluno: "))
            notaTeste = float(input("Indique a nota do Teste do aluno: "))
            aluno = (nome, id, [notaTPC, notaProj, notaTeste])
            turma.append(aluno)

    elif opção == "2":
        nome = input("Indique o nome do aluno: ")
        id = int(input("Indique o id do aluno: "))
        notaTPC = float(input("Indique a nota do TPC do aluno: "))
        notaProj = float(input("Indique a nota do Projeto do aluno: "))
        notaTeste = float(input("Indique a nota do Teste do aluno: "))
        aluno = (nome, id, [notaTPC, notaProj, notaTeste])
        if aluno not in turma:
            turma.append(aluno)
        else:
            print("Este aluno já existe na turma.")

    elif opção == "3":
        print("Lista de alunos na turma:")
        for aluno in turma:
            print(f"Nome: {aluno[0]}, ID: {aluno[1]}, Notas (TPC, Proj, Teste): {aluno[2]}")

    elif opção == "4":
        consulta_id = int(input("Indique o id do aluno que deseja consultar: "))
        encontrado = False
        for aluno in turma:
            if aluno[1] == consulta_id:
                print(f"Aluno encontrado:")
                print(f"Nome: {aluno[0]}")
                print(f"ID: {aluno[1]}")
                print(f"Notas (TPC, Proj, Teste): {aluno[2]}")
                encontrado = True
        if not encontrado:
            print("Aluno não encontrado na turma.")

    elif opção == "5":
        nome_ficheiro = input("Indique o nome do ficheiro onde deseja guardar a turma: ")
        guardar_turma_em_ficheiro(turma, nome_ficheiro)
        print("Turma guardada com sucesso.")

    elif opção == "6":
        nome_ficheiro = input("Indique o nome do ficheiro de onde deseja carregar a turma: ")
        turma = carregar_turma_de_ficheiro(nome_ficheiro)
        print("Turma carregada com sucesso.")

    elif opção == "0":
        print("A aplicação terminou.")
```
